```python
# jawaban nomor 3
umur = int(input("Masukkan umur: "))
punya_sim = input("Apakah memiliki SIM? (True/False): ") == "True"

# Operator logika AND
boleh_mengemudi = (umur >= 17) and punya_sim

# Contoh penggunaan tanda kurung untuk mengatur prioritas operasi
contoh_precedence = (umur >= 17) and (punya_sim or False)

print("Boleh mengemudi sendiri:", boleh_mengemudi)
print("Hasil ekspresi dengan tanda kurung:", contoh_precedence)
```

```python
# jawaban nomor 4
energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

# Menghitung efisiensi
efisiensi = (energi_output / energi_input) * 100

# Menampilkan hasil dengan f-string
print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")
```
